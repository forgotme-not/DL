```python
import numpy as np

# ===== 1. 创建数组 =====
a = np.zeros(3)                # (3,) 全 0 向量
b = np.ones((2, 3))            # 2行3列 全 1
c = np.full((200, 1), 0.01)    # 200行1列 全 0.01  ← 我们的 true_w 就是这种
d = np.random.randn(20, 200)   # 20行200列 标准正态随机数

# ===== 2. 形状 =====
d.shape                        # (20, 200)
d.T                            # 转置 -> (200, 20)

# ===== 3. 矩阵乘 @ vs 逐元素乘 * =====
X = np.random.randn(20, 200)   # 20 个样本, 200 维特征
W = np.random.randn(200, 1)    # 200x1 的权重列向量
X @ W            # 矩阵乘: (20,200)@(200,1) -> (20,1)  等价于 np.matmul(X, W)
#X * W            # 逐元素乘: 形状必须兼容,否则报错(广播规则下面讲)

# ===== 4. 广播:形状不同的数组自动对齐 =====
Z = X @ W                      # (20, 1)
Z + 0.05                       # 标量广播到每个元素  ← 我们的 true_b
Z + np.zeros(20)               # (20,1) + (20,) -> (20,1),向量也能广播

# ===== 5. 聚合 =====
np.mean(Z)                     # 所有元素均值 -> 标量
np.sum(Z)                      # 所有元素求和
np.sum(Z, axis=0)              # 沿"行方向"压缩 -> 每列的和,得到 (1,)
                               # axis=0 是竖着压缩,axis=1 是横着压缩

# ===== 6. 索引/切片(和列表一样)=====
X[:5]        # 前 5 行
X[:, :3]     # 所有行 + 前 3 列
c[:4]        # 前 4 个元素

# ===== 7. 最大的坑:(n,) 和 (n,1) 不是一回事 =====
v = np.array([1, 2, 3])        # 形状 (3,),一维"向量"
v.reshape(3, 1)                # (3, 1),二维"列矩阵"
v[:, None]                     # 等价写法,加一个轴
# 警告:v @ W 如果 W 是 (200,1) 而 v 是 (200,),会报错!
# 所以我们的代码里全程保持二维,比如 y 生成时就是 (n,1)
```
