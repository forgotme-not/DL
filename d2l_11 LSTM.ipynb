长短期记忆网络 LSTM










```python
import torch
import torch.nn as nn

dummy_input = torch.randn(1,5,3)#1个样本，5帧，3个特征

#开启双向：bidirectional=True
bilstm = nn.LSTM(input_size=3,hidden_size=8,batch_first=True,bidirectional=True)

out_bi,(h_n_bi,c_n_bi) = bilstm(dummy_input)
print("BiLSTM 的输出 out_bi 形状:", out_bi.shape)
```

```python
#取出最后一帧的特征
H_t_bi = out_bi[:,-1,:]

classifer_bi = nn.Linear(in_features=16,out_features=3)
```
