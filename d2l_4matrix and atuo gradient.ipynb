```python
#自动求导实现
import torch
x=torch.arange(4.0)
x
```

```python
x.requires_grad_(True) #等于 x=torch.arange(4.0,requires_grid=True)
x.grad
```

```python
y=2* torch.dot(x,x)
y
```

```python
y.backward()
x.grad
```

```python
x.grad == 4*x
```

```python
#在默认情况下，Pytorch会累积梯度，我们需要清除之前的值
x.grad.zero_()
y=x.sum()
y.backward()
x.grad
```

```python
x.grad.zero_()
y=x*x
y.sum().backward()
x.grad
```

```python
x.grad.zero_()
y=x*x
u=y.detach()
z=u*x

z.sum().backward()
x.grad==u
```

```python

x.grad.zero_()
y=x*x
u=y.detach()
z=u*x

z.sum().backward()
x.grad == u
```
