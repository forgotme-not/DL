```python
#内存相关
before =id(y)
y=y+x
id(y)==before  #更新值之后重新分配了地址
```

```python
z = torch.zeros_like(y)
print('id(z):',id(z))
z[:]=x+y  
print('id(z):',id(z))  #原地操作，没有改变内存
```
