```python
import numpy as np

# 一个单层神经网络

# 1. 输入数据
X = np.random.randn(100, 2)
# 修正 1：用 randint 生成 0 或 1 的标签
y = np.random.randint(0, 2, size=(100, 1))

# 2. 网络设置：权重矩阵和偏置
W = np.random.randn(2, 1) * 0.01
b = np.zeros((1,))

# 3. 用户交互（你写的防非法输入逻辑非常好！）
nums = int(input("输入训练轮次: "))
while nums <= 0:
    print("训练次数非法，请重新输入")
    nums = int(input("输入训练轮次: "))
    
lr = float(input("请输入学习率: "))
while lr <= 0:
    print("学习率非法，请重新输入")
    lr = float(input("请输入学习率: "))

# 4. 训练循环
for i in range(nums):
    # 修正 2：前向传播必须放在循环内部！每一轮拿新的 W 和 b 重新预测
    Z = X @ W + b
    #A = 1 / (1 + np.exp(-Z))
    A = np.maximum(0,Z)
    # 计算当前的 loss
    loss = np.mean((A - y) ** 2)

    # 修正 3：反向传播 (链式法则)
    dZ = 2 * (A - y) * A * (1 - A) / 100  # 结合了 loss求导 和 sigmoid求导，除以100取均值
    d_loss_W = X.T @ dZ                    # 矩阵乘法计算 W 的梯度 (2, 1)
    d_loss_b = np.sum(dZ, axis=0)          # 偏置 b 的梯度

    # 参数更新
    W -= lr * d_loss_W
    b -= lr * d_loss_b

    # 每 20 轮打印一次结果
    if i % 20 == 0:
        print(f"epoch: {i}, loss: {loss:.4f}")
```
